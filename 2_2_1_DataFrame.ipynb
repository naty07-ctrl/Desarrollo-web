Conclusiones del Análisis — Etapas de Comprensión del Negocio y de los Datos (CRISP-DM)

1. Trabajando con DataFrame
En esta sección se inicia el trabajo con estructuras de datos tipo DataFrame de Python, utilizadas en minería de datos para organizar, analizar y visualizar información de manera eficiente.

2. Subida de archivos al entorno de Google Colab
Se emplea el método files.upload() para cargar el archivo de datos desde el equipo local. Este paso permite disponer del conjunto de datos necesario para el análisis exploratorio dentro del entorno de Colab.

3. Importación de librerías
Se importan las librerías pandas y numpy, fundamentales para el tratamiento y análisis de datos.
pandas permite manipular estructuras tabulares (DataFrames), mientras que numpy ofrece soporte para operaciones numéricas y estadísticas de alto rendimiento.

4. Lectura del archivo de datos
Mediante pd.read_csv() se crea un DataFrame con tres variables principales: nota, género y asistencia.
Este paso organiza los datos en formato tabular, facilitando su comprensión y preparación para análisis posteriores.

5. Agrupación por género
El comando df_students.groupby("genero").size() muestra la cantidad de registros asociados a cada categoría de género.
Con esto se identifica la distribución de hombres y mujeres en la muestra, lo que aporta contexto al conjunto de datos.

6. Agrupación por nota
La agrupación por calificaciones (df_students.groupby('nota').size()) permite observar la frecuencia de cada nota obtenida por los estudiantes.
Este análisis aporta una primera visión sobre la dispersión y el rendimiento académico general.

7. Resumen estadístico del DataFrame
El método describe() entrega medidas como media, desviación estándar, valores mínimos y máximos.
Estas estadísticas básicas son esenciales para entender la tendencia central y la variabilidad de los datos disponibles.

8. Creación de un diccionario de géneros
Se genera un diccionario que asocia los códigos numéricos 0 y 1 con las etiquetas “Femenino” y “Masculino”.
Esto permite reemplazar valores numéricos por descripciones textuales, mejorando la comprensión de los datos.

9. Mapeo de los valores descriptivos
Se añade una nueva columna llamada desc_genero mediante el método map(), que asigna las descripciones de género a cada registro.
De esta forma, el DataFrame gana legibilidad y se prepara para su presentación o visualización final.

10. Visualización del DataFrame final
Se muestra el DataFrame completo, que ahora incluye la variable descriptiva del género junto con las notas y asistencias.
Este resultado refleja la culminación de las primeras etapas de la metodología CRISP-DM, correspondientes a la comprensión del negocio (conocer el contexto de los datos) y la comprensión de los datos (explorar, limpiar y preparar la información).
